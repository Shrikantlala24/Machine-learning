```python
class Stochastic_GD:
    def __init__(self, iterations=100, lr=0.1):
        self.w = None
        self.b = None
        self.iterations = iterations
        self.lr = lr
        self.loss_record = []

    def fit(self,X,y):
        Samples, features = X.shape

        # initializing parameters

        self.w = np.zeros(features)
        self.b = 0

        for epoch in range(self.iterations):
            for i in range(Samples):

                # pick one random sample
                # calculate y_pred - w and b
                # * but based on xi (one specific sample)
                # get residual (no loss needed)
                # get slope -  w and b
                # update    -  w and b

                idx = np.random.randint(0,Samples)
                xi = X[idx]
                yi = y[idx]

                y_pred = np.dot(xi,self.w) + self.b

                residual = yi - y_pred
                # no mse here needed
                
                slope_w = -2 * xi * residual
                slope_b = -2 * residual

                self.w -= self.lr * slope_w
                self.b -= self.lr * slope_b

            print(f"wieghts : {self.w} \nbiases : {self.b}")
```

```python
sgd = Stochastic_GD(100,0.1)
```
