```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
from sklearn.datasets import make_regression
# import pprint

X,y = make_regression(n_samples=10, n_features=5, n_informative=5, n_targets=1, random_state=42)

print(X)

print(f"\n----\n {y}")
```

```python
class MeraLR:
    def __init__(self, iterations, lr):
        self.w = None
        self.b = None
        self.iterations = iterations
        self.lr = lr
        self.loss_record = []

    def fit(self, X, y):
        samples, features = X.shape

        self.w = np.zeros(features)
        self.b = 0

        for i in range(self.iterations):

            # get y_pred ( = sample_dimension)
            # get residual ( = sample_dimension)
            # get mse_loss ( single value )
            # get slope of w and b
            # update w_new and b_new
            # append in loss_record

            y_pred = np.dot(X, self.w) + self.b

            residuals = y - y_pred

            mse = np.mean((y-y_pred)**2)

            n = len(y)

            slope_w = -2/n * np.dot(X.T, residuals)
            slope_b = -2/n * np.sum(residuals)

            self.w = self.w - self.lr*slope_w
            self.b = self.b - self.lr*slope_b

            print(f"for iteration {i} --- \nw = {self.w}\nb = {self.b}")

            # bUT HERE noticing part is that the entire X and y_actual is getting used and getting derivated to calc residual, mse, and  slopes
```

```python
lr_mera = MeraLR(100, 0.1)

```

```python
lr_mera.fit(X,y)
```
